```python
# Get 100 10-K forms' URLs
websitePrefix = "https://www.sec.gov/Archives/"
idxFile = "/content/drive/MyDrive/Text_Mining/company.idx" #在SEC网站上提前下好的文件

with open(idxFile) as indexFile:
    text = indexFile.readlines()

files = []

for line in text:
    parts = line.split()
    if len(parts) < 5:
        continue
    formType = parts[-4]
    if formType == "10-K":
        fileAttr = {
            "CIK": parts[-3],
            "URL": websitePrefix + parts[-1],
            "formType": formType,
            "fileDate": parts[-2],
        }
        files.append(fileAttr)
        if len(files) == 100:
            break

print("number of 10-K files:", len(files))
for fileAttr in files[:5]:
    print(fileAttr)
```

```python
#Download 100 10-K
import requests

headers = {
    "User-Agent": "TMclass 1340085@exa.com",
}

for fileAttr in files:
    url = fileAttr["URL"]
    response = requests.get(url, headers=headers, timeout=30)
    if response.status_code == 200:
        fileName = f"{fileAttr['formType']}_{fileAttr['CIK']}_{fileAttr['fileDate']}.txt"
        with open(downloadDir + fileName, "w") as f:
          f.write(response.text)
        print(f"Downloaded {fileName}")
    else:
        print(f"Skipped {url}: HTTP {response.status_code}")
```

```python
import re
import glob #用来找到文档位置
from collections import Counter

wordCounts = Counter()
textFiles = glob.glob(downloadDir + "*.txt")

for textFile in textFiles:
    with open(textFile, "r", encoding="utf-8", errors="ignore") as f:
        text = f.read().lower()

    words = re.findall(r"[a-z]+(?:'[a-z]+)?", text)
    wordCounts.update(words)

print(f"Counted {len(textFiles)} 10-K files")
print(f"Found {len(wordCounts)} unique words")

print(wordCounts.most_common(10))
```
