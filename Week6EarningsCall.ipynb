https://www.fool.com/earnings-call-transcripts/

https://www.fool.com/earnings-call-transcripts/page/10/

And download the 20 earning calls on each page.

For each earnings call:

- Split the call into parts: prepared remarks, and Q&As
- For each part, identify the speakers and their text (what they say)
- Store it all in two dataframes
  - DF1: [EarningsCallID, CompanyName, CallDate, EarningsCallText] - each row is an entire earnings call
  - DF2: [SpeakerTurnID (unique, sequential), EarningsCallID (never change from DF1),  SpeakerName, PreparedOrQA, Text] - each row is a speaker turn in a call

# Downloading 1-200 calls

```python
downloadDir = Path("/content/drive/MyDrive/Text_Mining/earnings_calls")
downloadDir.mkdir(parents=True, exist_ok=True)
```

```python
# collect 200 call links
callURLs = []

for page in range(1, 11):
  if page == 1:
    url = "https://www.fool.com/earnings-call-transcripts/"
  else:
    url = ("https://www.fool.com/earnings-call-transcripts/page/" + str(page) + "/")

  response = requests.get(url, timeout=30)
  response.raise_for_status()
  soup = BeautifulSoup(response.text, "html.parser")

  atags = soup.find_all("a", {"class": "text-gray-1100 group hover:no-underline!"})

  for a in atags:
    callURLs.append("https://www.fool.com" + a["href"])

  print("Page", page, ":", len(atags), "calls")
  time.sleep(1)
```

```python
print(len(callURLs)) # check if 200 call links
print(len(set(callURLs))) # check if any call links are duplicated
```

```python
# download each call and organize rows
callRows = []
speakerRows = []
needsReview = []

for callID, url in enumerate(callURLs, start=1):
  response = requests.get(url, timeout=30)
  response.raise_for_status()
  soup = BeautifulSoup(response.text, "html.parser")

  title = soup.find("h1").get_text(" ", strip=True)
  companyName = title.split(" (")[0]

  dateHeading = soup.find(id="date")
  dateText = dateHeading.find_next_sibling("p").get_text(" ", strip=True)
  callDate = pd.to_datetime(dateText).date().isoformat()

  transcriptHeading = soup.find(id="full-conference-call-transcript")
  paragraphs = transcriptHeading.find_next_siblings("p")

  paragraphTexts = []

  for p in paragraphs:
    paragraphTexts.append(p.get_text(" ", strip=True))

  callText = "\n\n".join(paragraphTexts)

  callRows.append({
      "EarningsCallID": callID,
      "CompanyName": companyName,
      "CallDate": callDate,
      "EarningsCallText": callText,
  })

  # Save the entire transcript as a text file.
  (downloadDir / f"call_{callID}.txt").write_text(callText, encoding="utf-8")

  section = "Prepared"
  currentTurn = None

  for p in paragraphs:
    text = p.get_text(" ", strip=True)
    bold = p.find("strong")

    # A bold label ending with ":" introduces a speaker.
    if bold and bold.get_text(strip=True).endswith(":"):
      label = bold.get_text(" ", strip=True)
      speakerName = label.rstrip(":").strip()
      speech = text[len(label):].strip()

      if (
        speakerName.lower() == "operator"
        and "first question" in speech.lower()
        ):
        section = "QA"

        currentTurn = {
            "SpeakerTurnID": len(speakerRows) + 1,
            "EarningsCallID": callID,
            "SpeakerName": speakerName,
            "PreparedOrQA": section,
            "Text": speech,
        }
        speakerRows.append(currentTurn)

      elif currentTurn is not None:
        # Continue the same speaker's turn across paragraphs.
        currentTurn["Text"] += "\n\n" + text

      else:
        raise ValueError(f"Missing initial speaker label: {url}")

    if section != "QA":
      needsReview.append((callID, url))

    print(callID, companyName, callDate)
    time.sleep(1)
```

```python
# save DFs
df1 = pd.DataFrame(callRows)
df2 = pd.DataFrame(speakerRows)

df1.to_csv(downloadDir / "DF1.csv", index=False)
df2.to_csv(downloadDir / "DF2.csv", index=False)
```
